# D3 Radial Equilibrium Equation

## Introduction

The pre-design studied in notebooks D1 and D2 give a first approach of the fan performance, focused on a 1D analysis. But they are not suitable for the complex flow produced
along the blade span in the axial fan. Specifically, it is expected to find a very different behavior near the hub and the tip of the fan, since the
actual flow in the turbomachine is three-dimensional.

The rotation of the flow downstream produces a **pressure gradient** along the radial direction, in order to balance the centrifugal force.
That leads to a redistribution of the axial 
velocity, $c_x$,
and, hence, of the flow rate. Besides, the pressure is also not uniform, and the performance of the fan has to be estimated by integrating both
the axial velocity and pressure. These two variables distributions are governed by the **Radial Equilibrium Equation (REE)**, that relates 
the derivatives on the radial equation with the profile of the _tangential velocity_, $c_\theta$, that is defined by the blade twisting and,
therefore, by the design of the fan.


## Learning Objectives

- **Understand** the importance of computing 3D distribution of velocity and pressure
- **Interpret** the Radial Equilibrium Equation (REE)
- **Solve the Inverse (Design) problem** by using the REE to determine the 3D blade design for a desired swirl distribution of flow.
- **Solve the Direct (Analysis) problem** by using the REE to infer fan performance from a blade design.
- **Identify** the physical onset of aerodynamics stall and flow reverse on off-design operating conditions.
